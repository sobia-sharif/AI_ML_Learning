```python
# map and filter #map changes every value
# if we want to9 add equal amounts in a number
# we use lambdfa for this.   #filter filters the specific things
# because lambda is preferable

#question add 1 to every number
numbers = [1,2,3,4]
list(map(lambda x: x++1, numbers))
```

```python
#question 2 multiply every number by 2
numbers = [5,6,7]
list(map(lambda x: x*2, numbers))
```

```python
#convert names to upper case
names = ["ali","sara", "john"]
list(map(lambda x: x.upper(),names))
```

```python
#keep numbers greater than 5
numbers = [2,5,8,10]
list(filter(lambda x: x>5, numbers))
```

```python
#keep even numbers
numbers=[12,2,3,4,5,6]
list(filter(lambda x : x%2 == 0, numbers))
```
